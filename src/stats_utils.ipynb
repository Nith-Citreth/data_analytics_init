```python
# Import des libraire

import pandas as pd
import numpy as np
```

```python
# Lecture du CSV et transformation en DataFrame

# on établis d'abord le chemin du fichier CSV
chemin = "C:/Users/Ossuaire/Desktop/data_analytics_init/data/sales.csv"
transactions = pd.read_csv(chemin)
```

```python
# Filtrage des valeurs du DataFrame pour enlever les valeurs nulles et négatives
transactions = transactions[transactions > 0]
transactions = transactions[transactions.notnull()]
```

```python
# Affichage du nombres de valeurs du DataFrame
print("Nombre de valeurs de transaction valide : ", len(transactions))
```

```python
# Affichage de la somme total des ventes
print("Somme total des ventes : ", transactions["montant"].sum())
```

```python
# Affichage de la moyenne des montants et de leurs nombres
print("Moyenne des montants : ", transactions["montant"].mean(), "pour un total de ", len(transactions["montant"]), "transactions")
```

```python
# Affichage de la médiane des montants
print("Médiane des montants : ", transactions["montant"].median())
```

```python
# Affichage de l'écart type des montants
print("Ecart type des montants : ", round(transactions["montant"].std(),2))
```

```python
# Affichage de la valeurs maximal et minimal des montants
print("Valeur maximale des montants : ", transactions["montant"].max())
print("Valeur minimale des montants : ", transactions["montant"].min())
```

```python
# Affichage des anomalies des montants (valeurs supérieur à 2 fois la moyenne), ainsi que leurs nombres
anomalies = transactions[transactions["montant"] > 2 * transactions["montant"].mean()]
print("Nombre d'anomalies : ", len(anomalies))
```

```python
# Création de la fonction analyser_ventes()
def analyser_ventes(transactions):
    # Filtrage des valeurs du DataFrame pour enlever les valeurs nulles et négatives
    transactions = transactions[transactions > 0]
    transactions = transactions[transactions.notnull()]

    # Affichage du nombres de valeurs du DataFrame
    print("Nombre de valeurs de transaction valide : ", len(transactions))

    # Affichage de la somme total des ventes
    print("Somme total des ventes : ", transactions["montant"].sum())

    # Affichage de la moyenne des montants et de leurs nombres
    print("Moyenne des montants : ", transactions["montant"].mean(), "pour un total de ", len(transactions["montant"]), "transactions")

    # Affichage de la médiane des montants
    print("Médiane des montants : ", transactions["montant"].median())

    # Affichage de l'écart type des montants
    print("Ecart type des montants : ", round(transactions["montant"].std(),2))

    # Affichage de la valeurs maximal et minimal des montants
    print("Valeur maximale des montants : ", transactions["montant"].max())
    print("Valeur minimale des montants : ", transactions["montant"].min())

    # Affichage des anomalies des montants (valeurs supérieur à 2 fois la moyenne), ainsi que leurs nombres
    anomalies = transactions[transactions["montant"] > 2 * transactions["montant"].mean()]
    print("Nombre d'anomalies : ", len(anomalies))
```

```python
# Affichage de l'index de transactions
print("Index de transactions : ", len(transactions["id"]))
```

```python
analyser_ventes(transactions)
```
